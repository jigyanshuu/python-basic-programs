```python
def convertbinary(n):
        if n==0:
                return "0"
        res=""
        while n!=0:
                if n%2==1:
                        res+="1"
                else:
                        res+="0"
                n=n//2
        return res[::-1]
a=convertbinary(13)
print(a)
```

```python
def convertDecimal(x):
        length=len(x)
        p2=1
        num=0
        for i in range(length-1, -1, -1):
                if x[i]=='1':
                        num+=p2
                p2=p2*2
        return num
a=convertDecimal("1101")
print(a)
```

```python
def solve(s, p):
        if sorted(s)==sorted(p):
                return True
        else:
                return False
s=input()
p=input()
if solve(s, p):
        print("YES")
else:
        print("NO")
```

```python
def solve1(s, p):
        s1=s.split()
        p1=p.split()
        s1.sort()
        p1.sort()
        if s1==p1:
                return True
        else:
                return False
s=input()
p=input()
if solve1(s, p):
        print("Yes")
else:
        print("NO")
```
